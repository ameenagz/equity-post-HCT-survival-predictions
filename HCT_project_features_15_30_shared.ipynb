Loading in data

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

test = pd.read_csv("/content/test.csv")
print("Test shape:", test.shape )

train = pd.read_csv("/content/train.csv")
print("Train shape:",train.shape)
train.head()
```

Handling features

```python
remove = ["ID"]
features = [c for c in train.columns if not c in remove]
print(f"There are {len(features)} FEATURES: {features}")
```

```python
#divide 57 features into 4 groups
featuresA = features[:15]
featuresB = features[15:30]
featuresC = features[30:45]
featuresD = features[45:]
print(f"There are {len(featuresA)} FEATURES: {featuresA}")
print(f"There are {len(featuresB)} FEATURES: {featuresB}")
print(f"There are {len(featuresC)} FEATURES: {featuresC}")
print(f"There are {len(featuresD)} FEATURES: {featuresD}")
```

```python
catFeaturesB = []
for c in featuresB:
  if train[c].dtype == "object":
    catFeaturesB.append(c)
print(f"There are {len(catFeaturesB)} CATEGORICAL FEATURES: {catFeaturesB}")

numFeaturesB = [c for c in featuresB if not c in catFeaturesB]
print(f"There are {len(numFeaturesB)} NUMERICAL FEATURES: {numFeaturesB}")
```

**NULL Val INTERPRETATION IN EACH FEATURE B**

```python
for c in catFeaturesB:
  print(train[c].value_counts(dropna=False))
```

```python
#correlation of missing values with efs
for c in catFeaturesB:
  missing_c = train[c].isnull()
  missing_impact = train.groupby("efs")[c].apply(lambda x: x.isnull().sum())
  print(missing_impact)
```

**Handling NON-Signficant columns**

tce_imm_match,  rituximab,  prod_type,  ethnicity,  obesity

```python
#filling in null values appropriately
train["tce_imm_match"] = train["tce_imm_match"].fillna("Missing")
train["ethnicity"] = train["ethnicity"].fillna("Missing")
train["obesity"] = train["obesity"].fillna("Not done")

#using mode frequency
train["rituximab"] = train["rituximab"].fillna(train["rituximab"].mode()[0])
```

```python
#Encoding Categorical Features

from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
#one hot encoding
train = pd.get_dummies(train, columns=["tce_imm_match", "ethnicity"], dtype=int)
train = pd.get_dummies(train, columns=["prod_type"], dtype=int)
train = pd.get_dummies(train, columns=["obesity"], dtype=int)

#Binary encoding
train["rituximab"] = train["rituximab"].map({"Yes": 1, "No": 0})
```

```python
  #CheckPoint
columns_to_check = ["tce_imm_match", "rituximab", "prod_type", "ethnicity", "obesity"]

for col in columns_to_check:
    encoded_cols = [c for c in train.columns if col in c]  # Find all one-hot encoded variants
    print(f"\n--- Checking: {col} ---")
    print(train[encoded_cols].head())  # Print first 5 rows of one-hot encoded columns
    print(f"Sum of 'Missing' for {col}: {train[[c for c in encoded_cols if 'Missing' in c]].sum().sum()}")  # Count missing values
```

**Handling Significant columns**

cmv_status, conditioning_intensity, cyto_score_detail

 treating missing values as a separate category can help capture any unique characteristics of those patients. This approach acknowledges that missing data might indicate a different treatment path or patient experience, which could be relevant to understanding disparities

```python
train["cmv_status"]= train["cmv_status"].fillna("Missing")
train["conditioning_intensity"]= train["conditioning_intensity"].fillna("Missing")
train["cyto_score_detail"]= train["cyto_score_detail"].fillna("Missing")
```

```python
#Ordinal Encoding cmv_status:

cmv_mapping = {
    "-/+": 3,      #Highest risk
    "+/-": 2,
    "-/-": 1,
    "+/+": 0,      #Lowest risk
    "Missing": -1  #category for missing
}

train["cmv_status"] = train["cmv_status"].map(cmv_mapping)
```

```python
cytoscore_mapping = {
    'Poor': 4,                              # Highest risk
    'TBD': 3,
    'Not tested': 2,
    'Intermediate': 1,
    'Favorable': 0,                         # Lowest risk
    'Missing': -1                           # Category for missing
}
train["cyto_score_detail"] = train["cyto_score_detail"].map(cytoscore_mapping)
```

```python
conditioning_mapping = {
    'MAC': 5,                              # Lowest intensity
    'RIC': 4,
    'NMA': 3,
    'TBD': 2,
    'No drugs reported': 1,
    'N/A, F(pre-TED) not submitted': 0,    #Highest intesity
    'Missing': -1                          # Category for missing
}
train["conditioning_intensity"] = train["conditioning_intensity"].map(conditioning_mapping)
```

```python
      #Check Point:
original_unique_values = train['cyto_score_detail'].unique()
print("Original unique values in cyto_score_detail:", original_unique_values)

original_unique_values = train['cmv_status'].unique()
print("Original unique values in cmv_status:", original_unique_values)

original_unique_values = train['conditioning_intensity'].unique()
print("Original unique values in conditioning_intensity:", original_unique_values)
```

```python
# make a copy of train with just the encoded data
cols_to_add = ['cmv_status','cyto_score_detail','conditioning_intensity','rituximab','tce_imm_match_G/B', 'tce_imm_match_G/G', 'tce_imm_match_H/B', 'tce_imm_match_H/H', 'tce_imm_match_Missing', 'tce_imm_match_P/B', 'tce_imm_match_P/G', 'tce_imm_match_P/H', 'tce_imm_match_P/P', 'ethnicity_Hispanic or Latino', 'ethnicity_Missing', 'ethnicity_Non-resident of the U.S.', 'ethnicity_Not Hispanic or Latino', 'prod_type_BM', 'prod_type_PB', 'obesity_No', 'obesity_Not done', 'obesity_Yes']
train1 = train[cols_to_add].copy()
train1.head()
```

**Handling Numerical Features**

```python
numFeaturesB = [c for c in featuresB if not c in catFeaturesB]
print(f"There are {len(numFeaturesB)} NUMERICAL FEATURES: {numFeaturesB}")
```

```python
hla_features = ['hla_high_res_10', 'hla_match_dqb1_high', 'hla_match_c_low', 'hla_match_dqb1_low']

for feature in hla_features:
    median_value = train[feature].median()
    train[feature] = train[feature].fillna(median_value)
```

```python
nonsig_hla_features = ['hla_match_drb1_low', 'hla_nmdp_6']

for feature in nonsig_hla_features:
  median_value = train[feature].median()
  train[feature] = train[feature].fillna(median_value)
```
