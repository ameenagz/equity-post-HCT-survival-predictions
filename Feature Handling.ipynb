mrd_hct to gvhd_proph

```python
def encode_columns(column):
  #replace NaN with NA
  column_filled = column.fillna('NA').astype(str)  #for missing values

  #Applying custom mapping
  custom_mapping = {  #positive/yes = 1 .. negative/one = 0
      'Positive' : 1,
      'Negative' : 0,
      'Yes' : 1,
      'No' : 0,
      'Not done' : 2,
      'NA' : -1  #missing value
  }
  column_filled = column_filled.map(custom_mapping)
  return column_filled
```

tce_match, sex_match, gvhd_proph have more than just the basic categorical values so applying mapping individually for each

tce_match

```python
#starting with tce_match
def encode_columns(column):
  column_filled = column.fillna('NA').astype(str)

  custom_mapping = {
      'Permissive' : 0,
      'Fully matched' : 1,
      'GvH non-permissive' : 2,
      'HvG non-permissive' : 3,
      'NA' : -1
  }
  column_filled = column_filled.map(custom_mapping)
  return column_filled

train_df['tce_match'] = encode_columns(train_df['tce_match'])
```

sex_match

```python
#sex_match
def encode_columns(column):
  column_filled = column.fillna('NA').astype(str)

  custom_mapping = {
      'F-F' : 1,
      'M-M' : 2,
      'M-F' : 3,
      'F-M' : 4,
      'NA' : -1
  }

  column_filled = column_filled.map(custom_mapping)
  return column_filled

train_df['sex_match'] = encode_columns(train_df['sex_match'])
```
